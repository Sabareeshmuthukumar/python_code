```python
#matrix addition
rows = int(input("Enter number of rows: "))
cols=int(input("Enter number of cols :"))
matrix1 = []
matrix2=[]
result=[]
for i in range(rows):
    row = []
    for j in range(cols):
        row.append(int(input("enter the values :")))
    matrix1.append(row)

for i in range(rows):
    row = []
    for j in range(cols):
        row.append(int(input("enter the values :")))
    matrix2.append(row)

for i in range(rows):
    row = []
    for j in range(cols):
        total = matrix1[i][j] + matrix2[i][j]
        row.append(total)
    result.append(row)

print(matrix1)
print(matrix2)
print(result)
```

```python
#trapping rain water
height=[0,1,0,1]
n=len(height)
if n==0:
    print(0)
else:
    left_max=[0]*n
    left_max[0]=height[0]

    right_max=[0]*n
    right_max[n-1]=height[n-1]

    for i in range(1,n):
        left_max[i]=max(left_max[i-1],height[i])

    for i in range(n-2,-1,-1):
        right_max[i]=max(right_max[i+1],height[i])

    water=0
    for i in range(n):
        water+=min(left_max[i],right_max[i])-height[i]

    print(water)
```

```python
#sort a list
l=[]
n=int(input())
for i in range(n):
    l.append(int(input()))

print(l)
def bubble_sort(l):
    for i in range(n-1):
        for j in range(n-i-1):
            if l[j]>l[j+1]:
                temp=l[j]
                l[j]=l[j+1]
                l[j+1]=temp
    return l

print(bubble_sort(l))
```

```python
#finding even and odd also find their min max and average
l=[]
e=[]
o=[]
n=int(input())
for i in range(n):
    l.append(int(input()))

print(l)

for i in range(n):
    if l[i]%2==0:
        e.append(l[i])
    elif l[i]%2!=0:
        o.append(l[i])

print("Even",e)
print("Odd",o)

print("min of even",min(e))
print("min of odd",min(o))

print("max of even",max(e))
print("max of odd",max(o))

print("average of even",(sum(e)/len(e)))
print("average of odd",(sum(o)/len(o)))

```
